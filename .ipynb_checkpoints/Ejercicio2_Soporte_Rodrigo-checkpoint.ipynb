## Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | tiempo_espera_minutos, nivel_urgencia, cliente_premium |
| **Entorno (E)** | Mesa de ayuda que recibe tickets de distintos clientes de forma simultánea |
| **Acciones (A)** | Resolver en nivel 1, escalar a nivel 2, escalar a nivel 3 |
| **Objetivo** | Atender primero los tickets más urgentes sin dejar esperando demasiado a nadie |
| **Medida de desempeño (P)** | Tiempo de resolución, satisfacción del cliente, uso balanceado de los niveles |

### Justificación de las reglas ###

La urgencia alta se escala de inmediato porque suele tratarse de fallas graves que no pueden esperar. La urgencia media depende del tiempo de espera porque, aunque no es crítica al inicio, si el cliente ya lleva esperando mucho, el problema se vuelve más urgente. Se le da prioridad extra a los clientes premium porque el banco (o la empresa) tiene un compromiso de atención más rápido con ellos, aunque esto no debe hacer que se ignore a un cliente normal con urgencia alta.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    if nivel_urgencia == "media":
        limite = 15 if cliente_premium else 30
        if tiempo_espera_minutos >= limite:
            return "escalar a nivel 2", f"urgencia media con espera >= {limite} min"
        return "resolver en nivel 1", "urgencia media dentro del tiempo tolerable"

    # urgencia baja
    if cliente_premium and tiempo_espera_minutos >= 45:
        return "escalar a nivel 2", "cliente premium con espera prolongada pese a baja urgencia"
    return "resolver en nivel 1", "urgencia baja sin condiciones que ameriten escalar"
```

### Simulación y pruebas

```python
casos = [
    (5, "baja", False),
    (10, "alta", False),
    (20, "media", False),
    (10, "media", True),
    (50, "baja", True),
    (15, "media", True),   # caso limite: justo en el margen para premium
]

print(f"{'Espera':>7} | {'Urgencia':>8} | {'Premium':>7} | Acción / Motivo")
print("-" * 65)
for espera, urgencia, premium in casos:
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"{espera:>7} | {urgencia:>8} | {str(premium):>7} | {accion} — {motivo}")
```
